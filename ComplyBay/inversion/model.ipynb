```python
import numpy as np
from ComplyBay.inversion.model import VpEstimate, RhoEstimate
import matplotlib.pyplot as plt
```

```python
d = [0, 500, 1000, 2000, 3500, 5000]
z = np.array(d) * 1e-3  # Convert to km
vs = np.array([0.1, 0.8, 1.5, 2.7, 2.8, 3.0])  # Example shear wave velocities in km/s

vp_estimator = VpEstimate(z=z, vs=vs)
rho_estimator = RhoEstimate(z=z, vs=vs)
```

```python
fig, ax = plt.subplots(1, 3, sharey=True, figsize=(10, 5))
ax[0].plot(vs, z, label='Vs')

for keyword in ["Brocher2005", "Castegna1995"]:
    vp = vp_estimator.estimate(keyword=keyword)
    ax[1].plot(vp, z, label=f'Vp {keyword}')

for keyword in ["Brocher2005", "Willoughby&Edwards2000"]:
    rho = rho_estimator.estimate(keyword=keyword)
    ax[2].plot(rho, z, label=f'Rho {keyword}')

# ax[1].set_ylim(0, 2)
ax[1].legend()
ax[2].legend()
ax[0].invert_yaxis()
```
